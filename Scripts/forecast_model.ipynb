## 4. Investigating outliers

```python
# Create a numeric columns variable
for name, dataset in datasets.items():
    numeric_variables = dataset.select_dtypes(include="number")
    print(f"\n{name}")
    
    if len(numeric_variables.columns):
        display(numeric_variables.describe().T)
    else:
        print("No numeric columns in dataset.")
```

```python
# Using the IQR rule to check outliers

all_outliers = []
for name, dataset in datasets.items():
    numeric_variables = dataset.select_dtypes(include="number")

    for column in numeric_variables.columns:
        data = numeric_variables[column].dropna()
        Q1 = data.quantile(0.25)
        Q3 = data.quantile(0.75)
        IQR = Q3-Q1
        lower_limit = Q1 - (IQR * 1.5)
        upper_limit = Q3 + (IQR * 1.5)

        outliers = data[(data < lower_limit)|(data > upper_limit)]
        all_outliers.append({"Dataset": name,
                            "Column": column,
                            "Outliers(%)": round(len(outliers)/len(data)*100, 2)})
        
display(pd.DataFrame(all_outliers))
```

## 5. Feature Engineering

- The model forecasts weekly and monthly forecasts for the 3 targets, grouped by store branch and product category: **sales revenue**, **units sold**, and **transaction volume**.
- Selected features include: year, month, end_month_indicator, store branch, product category, store floor area, promotion indicator, discount percentage and inflation rate.
- Targets use ordinary sales, therefore returns/refunds are not included
-Revenue is defined as `quantity * unit price * (1-discount/100)`


### Time Features

```python
# Extracting time-based features from transaction dates
transactions["year"] = transactions["transaction_date"].dt.year
transactions["month"] = transactions["transaction_date"].dt.month
transactions["week"] = transactions["transaction_date"].dt.isocalendar().week
transactions["day_name"] = transactions["transaction_date"].dt.day_name()
transactions["end_month"] = transactions["transaction_date"].dt.is_month_end
```

### Preparing sales data

```python
#Excluding returns, they are not included in the model
sales = receipt_lines[receipt_lines["quantity"]>0].copy() #new sales df

#Adding transaction dates, store and time features
sales = sales.merge(transactions[["transaction_id", "transaction_date","year", "month", 
                                  "week","day_name","end_month",  "store_id"]], 
                                  on = "transaction_id", how = "left")
# Adding product category
sales = sales.merge(products[["product_id", "category"]], 
                    on = "product_id", how = "left")

# Adding a revenue column
sales["revenue_kes"] = (sales["quantity"] * sales["unit_price_kes"] * 
                        (1-sales["discount_pct"] /100))

# To be able to do weekly forecast, we need to determine start of each week
sales["week_start"] = (sales["transaction_date"]-
                       pd.to_timedelta(sales["transaction_date"].dt.day_of_week, unit = "D"))

# Identify sales linked to a promo
sales["promo_indicator"] = sales["promo_id"].notna().astype(int)

# Identifying sales linked to a holiday
sales["is_holiday"] = sales["transaction_date"].isin(holidays["date"]).astype(int)
```

### Aggregating sales - To enable weekly and monthly forecasting


1. Weekly

```python
# Aggregate sales by week, store and product category.
weekly_data = sales.groupby(
    ["week_start", "store_id", "product_id", "category"], as_index = False).agg(
       revenue_kes=("revenue_kes", "sum"), units_sold=(
          "quantity", "sum"),transaction_count=("transaction_id", "nunique"),promotion_indicator=(
             "promo_indicator", "max"), is_holiday=("is_holiday", "max"), discount_pct=("discount_pct", "mean"))

# Adding time features to weekly dataset

weekly_data["year"] = weekly_data["week_start"].dt.year
weekly_data["month"] = weekly_data["week_start"].dt.month

week_end = weekly_data["week_start"] + pd.Timedelta(days=6)
weekly_data["end_month_indicator"] = (week_end.dt.is_month_end
    | (weekly_data["week_start"].dt.month != week_end.dt.month)).astype(int)

# Adding the inflation rate based on the week start month.
inflation_by_month = inflation.set_index(
   inflation["month"].dt.to_period("M"))["inflation_12m_pct"]

weekly_data["inflation_12m_pct"] = weekly_data[
   "week_start"].dt.to_period("M").map(inflation_by_month)

# Adding lag features (For previous week sales)
weekly_data = weekly_data.sort_values(["store_id", "product_id", "week_start"])

for column in ["revenue_kes", "units_sold", "transaction_count"]:
   weekly_data[f"previous_{column}"] = (weekly_data.groupby(
      ["store_id", "product_id"])[column].shift(1))
```

2. Monthly

```python
# Assigning each sale to month.
sales["month_start"] = (sales["transaction_date"].dt.to_period("M").dt.to_timestamp())

monthly_data = sales.groupby(
    ["month_start", "store_id", "product_id", "category"],as_index = False).agg(
       revenue_kes=("revenue_kes", "sum"), units_sold=(
           "quantity", "sum"),transaction_count=("transaction_id", "nunique"),promotion_indicator=(
               "promo_indicator", "max"), is_holiday=("is_holiday", "max"), discount_pct=("discount_pct", "mean"))

# Adding time features 
monthly_data["year"] = monthly_data["month_start"].dt.year
monthly_data["month"] = monthly_data["month_start"].dt.month

# Add the monthly inflation ratate.
inflation_by_month = inflation.set_index(
    inflation["month"].dt.to_period("M"))["inflation_12m_pct"]
monthly_data["inflation_12m_pct"] = monthly_data[
    "month_start"].dt.to_period("M").map(inflation_by_month)

# Adding lag features (For previous week sales)
monthly_data = monthly_data.sort_values(["store_id", "product_id", "month_start"])
for column in ["revenue_kes", "units_sold", "transaction_count"]:
    monthly_data[f"previous_{column}"] = (
        monthly_data.groupby(["store_id", "product_id"])[column].shift(1))
```

```python
#Arranging and viewing weekly columns
feature_columns = ["year", "month", "end_month_indicator",
    "store_id", "product_id", "category", "promotion_indicator", 
    "discount_pct", "inflation_12m_pct", "is_holiday","previous_revenue_kes", 
    "previous_units_sold","previous_transaction_count"]

target_columns = ["revenue_kes", "units_sold", "transaction_count"]
weekly_data = weekly_data[["week_start"] + feature_columns + target_columns]

weekly_data.head()
```
